```python
""" a school calculated iq score of 50 students, the average turned to be 100.
the mean of population iq is 90 and the std devation 16, state whether the claim by scholl that iq increases
if student study more then avg of scholl with 5% of significance level."""

#null hypothesis : mean_iq = 90
#alternale hypothesis : mean_iq >90
```

```python
import scipy.stats as stats   #z-critical approach

sample_mean = 100
population_mean = 90
population_std = 16
sample_size = 60
alpha = 0.05
```

```python
zscore = (sample_mean - population_mean)/(population_std/np.sqrt(50))
print("the zscore is :",zscore)
```

```python
#stats >> ststistics module in scipy
#norm >> nomal distribulion 
#ppf >> percent point function > inverse of cumulative distribulion function(cdf)

z_critical = stats.norm.ppf(1 - alpha)
```

```python
z_critical
```

```python
 #z-critical approach
if zscore > z_critical:
    print("reject the null hypothesis")
else:
    print("fail to reject the null hypothesis")
```

```python
#p-value approch
p_valu = 1-stats.norm.cdf(zscore)

if p_valu < alpha:
    print("reject the null hypothesis")
else:
    print("fail to reject the null hypothesis")
```

```python
observed_values = data_table.values
```

```python
st_test, pvalue, dof, expected_values = stats.chi2_contingency(observed_values)
```

```python
#chi square test >> observed_valus and expected_values -- useing critical method

expected_values
```

```python
observed_values
```

```python
# dof = (no of row -1)*(no of columns -1)

dof = 1

from scipy.stats import chi2
```

```python
for i in zip(observed_values, expected_values):
    print(i)
```

```python
chisquare_test = sum([(o-e)**2 / e for o, e in zip(observed_values, expected_values)])
```

```python
chisquare_test
```

```python
chisquare_statistics = chisquare_test[0] + chisquare_test[1]
chisquare_statistics
```

```python
dof,alpha
```

```python
chi2_critical = chi2.ppf(1-alpha, df = dof)
```

```python
chi2_critical
```

```python
if chisquare_statistics >= chi2_critical:
    print("reject the null hypothesis, there is relationship between two catagorical variable")
else:
    print("fail to reject the null hypothesis, no relationship")
```

```python
# p valu mathod

p_value = 1-chi2.cdf(x = chisquare_statistics, df = dof)

if p_value <=alpha:
    print("reject the null hypothesis, there is relationship between two catagorical variable")
else:
    print("fail to reject the null hypothesis, no relationship")
```

```python
# F test

# example : tow workers >> if one worker is effective more then other one

alpha = 0.05
worker1 = [18, 17, 20, 25, 27, 55, 51, 42, 46, 29]
worker2 = [18, 16, 22, 26, 39, 51, 48, 44, 58, 12]
```

```python
#calculate f-statistics 
f_stat = np.var(worker1)/np.var(worker2)
f_stat
```

```python
df1 = len(worker1)-1
df2 = len(worker2)-1
```

```python
critical_valu = stats.f.ppf(q = 1-alpha, dfn = df1, dfd = df2)
```

```python
critical_valu
```

```python
if f_stat>critical_valu:
    print("reject the null hypothesis")
else:
    print("fail to reject the null hypothesis")
```
