```python
#relplot to make a line plot >> we generated a time-serise data
value = np.random.randn(500).cumsum()
time = np.arange(500)
df = pd.DataFrame({'time':time, 'value':value})
```

```python
df
```

```python
sns.relplot(x='time', y='value', kind='line', data=df, sort=True)
```
