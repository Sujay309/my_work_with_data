```python
#write a code to interchange first and last element in the list 
def swap_first_last(lst):
    lst[0], lst[-1] = lst[-1], lst[0]  # Swapping the first and last elements
    return lst
my_list = [1, 2, 3, 4, 5]
swap_first_last(my_list)
print(my_list)  
```

```python
password = input("enter the password :")
def valid_pass(password):
    upper = False
    lower = False
    digit = False
    symbol = False
    symbol_list=" / . , ; ' [ ] = - ) ( * & ^ % $ # @ ! ~ < > ? : | "
    if len(password) >8 and len(password) <20:
        for i in password:
            if i.isupper():
                upper = True
            elif i.islower():
                lower =True
            elif i.isdigit():
                digit = True
            elif i in symbol_list():
                symbol = True
if valid_pass(password):
    print("the password saved")
else:
    print("invalid password!")
```
