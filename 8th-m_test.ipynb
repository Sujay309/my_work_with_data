```python
x =[1,2,3,4,5]
y =[5,1,7,8,2]
plt.plot(x,y)
```

```python
plt.plot(x,y, color ='red', marker = "o", linestyle ='--')
plt.title("line plot to understand the sentex")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.plot(x,y, color ='red', marker = "o", linestyle ='--', linewidth = 3)
plt.title("line plot to understand the sentex")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.plot(x,y, color ='red', marker = "o", linestyle ='--', linewidth = 3, markersize = 10)
plt.title("line plot to understand the sentex")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.plot(x,y, color ='red', marker = "o", linestyle ='--', linewidth = 3)
plt.title("line plot to understand the sentex")
plt.xlabel("x points")
plt.ylabel("y points")
plt.grid()
plt.show()
```

```python
x =[1,2,3,4,5]
y1 =[5,1,7,8,2]
y2 = [7,8,3,5,1]

plt.plot(x,y1, color ='r', marker = "o", linestyle ='--', linewidth = 3, markersize = 6, label = 'line y1')
plt.plot(x,y2, color ='b', marker = "o", linestyle ='--', linewidth = 3, markersize = 6, label ='line y2')
plt.xlabel("x data exis")
plt.ylabel("y data exis")
plt.title("multiple line in one plot")
plt.legend()
plt.show()
```

```python
#piechart >> to understand the percen distribution on dat

langs = ['c','python','java']
student =[20,100,40]

plt.pie(student, labels=langs, autopct='%1.1f%%')
plt.title("piechart")
plt.show()
```

```python
explode = (0.0,0.1,0.1)
langs = ['c','python','java']
student =[20,100,40]

plt.pie(student, labels=langs, autopct='%1.1f%%', explode=explode)
plt.title("piechart")
plt.show()
```

```python
explode = (0.3,0.1,0.2)
langs = ['c','python','java']
student =[20,100,40]

plt.pie(student, labels=langs, autopct='%1.1f%%', explode=explode, shadow=True)
plt.title("piechart")
plt.show()
```
